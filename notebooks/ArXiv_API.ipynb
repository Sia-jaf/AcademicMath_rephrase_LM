Look at the following link for arxiv module [link](https://pypi.org/project/arxiv/).

```python
import arxiv
import os
import tarfile

# 1. Define the arXiv paper ID you want to download
#    Example ID: "1706.03762" (the version can also be specified like "1706.03762v1")
paper_id = "1706.03762"

# 2. Use the arxiv client to search for the paper
client = arxiv.Client()
search = arxiv.Search(id_list=[paper_id])
results = list(client.results(search))

if not results:
    print(f"Paper with ID {paper_id} not found.")
else:
    # Get the first (and likely only) result
    paper = results[0]
    
    # 3. Define the directory where you want to save the source file
    download_dir = "../data/arxiv_sources"
    os.makedirs(download_dir, exist_ok=True)
    
    # 4. Download the source package
    # The download_source method saves the file as a .tar.gz archive (often without an extension)
    print(f"Downloading source for {paper.title}...")
    source_path = paper.download_source(dirpath=download_dir, filename=f"{paper_id}.tar.gz")
    print(f"Source downloaded to: {source_path}")
    
    # 5. Extract the LaTeX files from the downloaded archive
    try:
        with tarfile.open(source_path, "r:gz") as tar:
            tar.extractall(path=download_dir)
        print(f"Source extracted to: {download_dir}")
        # Optional: remove the archive after extraction
        # os.remove(source_path) 
    except tarfile.ReadError:
        print("Could not extract archive. The source might be a single .tex file.")
        # If it's a single .tex file, the downloaded file will be the file itself (without tar.gz structure)
    except Exception as e:
        print(f"An error occurred during extraction: {e}")

```
