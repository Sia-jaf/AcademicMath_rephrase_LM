**RegEx Functions**

The re module offers a set of functions that allows us to search a string for a match:

|Function | Description|
|:--------|:-----------|
|findall |Returns a list containing all matches|
|search |Returns a Match object if there is a match anywhere in the string|
|split |Returns a list where the string has been split at each match|
|sub |Replaces one or many matches with a string|

For a reference look at the [Python re Module Documentation](https://docs.python.org/3/library/re.html), [RegEx Cheatsheet](https://www.debuggex.com/cheatsheet/regex/python), [RegEx Tutorial](https://www.w3schools.com/python/python_regex.asp) on W3Schools. Lastly, [Gitup repo](https://learnbyexample.github.io/python-regex-cheatsheet/#understanding-python-re-gex-book).
